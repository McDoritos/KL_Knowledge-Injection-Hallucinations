```python
from ollama import chat
from ollama import ChatResponse
from ollama import embed
import json
import glob
from tqdm import tqdm
import os
```

```python
GENERATION_MODEL = "qwen3:4b" # qwen3:4b or qwen3:1.7b
```

```python
file = "../Knowledge-graph/doc_146120936/questions_146120936.txt"

all_questions = []

with open(file, "r", encoding="utf-8") as f:
    qs = [q.strip() for q in f.readlines() if q.strip()]
    all_questions.extend(qs)

print(f"Total questions found: {len(all_questions)}\n")

doc_folder = os.path.dirname(file)
doc_name = os.path.basename(doc_folder)
output_file = os.path.join(doc_folder, f"Base-Output-4B_{doc_name}.json")

pbar = tqdm(total=len(all_questions), desc="Processing questions")

results = {"results": []}

print(f"\nProcessing file: {file}\n")

with open(file, "r", encoding="utf-8") as f:
    questions = f.readlines()

for question in questions:
    if question.strip() == "":
        continue

    pbar.update(1)
    question = question.strip()

    print(f"\n===== Question =====\n{question}\n")

    # PROMPT (no context, no RAG wording)
    prompt = f"""
You are an information extraction system trained to produce 
Knowledge Graph (KG) triplets following the SciERC schema.

========================================
TASK
========================================
Given ONLY the research question below, extract all plausible factual
knowledge graph triplets that can be reasonably inferred from the question.

1. Triplet format:
[subject:ENTITY_LABEL, RELATION_LABEL, object:ENTITY_LABEL]

2. Allowed ENTITY_LABEL values:
- Method
- Task
- Dataset

3. Allowed RELATION_LABEL values:
- Used-For
- Part-Of
- Compare-With
- SubClass-Of
- Synonym-Of
- Evaluated-With
- Benchmark-For
- Trained-With
- SubTask-Of

4. Output rules:
- ALWAYS output valid JSON containing an array of triplets.
- DO NOT add explanations or text outside the JSON.
- DO NOT invent arbitrary unrelated entities.
- You MAY infer relationships implied by the question itself.
- If multiple triplets seem valid, output all of them.
- The output must contain ONLY the JSON object described below.

========================================
QUESTION
{question}

========================================
OUTPUT FORMAT
Return ONLY a JSON object with the following structure:
{{
"triplets": [
"[subject:LABEL, RELATION, object:LABEL]"
]
}}

Do NOT add any text before or after the JSON.
"""

    stream = chat(
        model=GENERATION_MODEL,
        messages=[{"role": "user", "content": prompt}],
        stream=True,
    )

    llm_output = ""
    for chunk in stream:
        llm_output += chunk["message"]["content"]

    print(chunk["message"]["content"], end="", flush=True)
    print("\n")

    try:
        triplets_json = json.loads(llm_output)
    except:
        triplets_json = {"triplets": [], "raw_output": llm_output}

    results["results"].append({
        "question": question,
        "triplets": triplets_json
    })

    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(results, f, indent=4, ensure_ascii=False)

    print(f"\nSaved output to: {output_file}\n")

pbar.close()
```
